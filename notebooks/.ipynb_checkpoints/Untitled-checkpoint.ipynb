```python
🟡 CELL 1 — Imports
```

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             classification_report, roc_curve)
from imblearn.over_sampling import SMOTE
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from collections import Counter
import joblib, os, time
from scipy.stats import randint

print("✅ All libraries imported!")
```

```python
🟡 CELL 2 — Load Dataset
```

```python
df = pd.read_csv(r'C:\Users\soumya\OneDrive\Desktop\Credit_Card_Fraud_Detection\data\creditcard.csv')

print("="*50)
print(f"Shape:         {df.shape}")
print(f"Missing vals:  {df.isnull().sum().sum()}")
print(f"Duplicates:    {df.duplicated().sum()}")
print("="*50)
print("Class Distribution:")
print(df['Class'].value_counts())
print(f"\nFraud %: {df['Class'].mean()*100:.4f}%")
print("="*50)
df.head()
```

```python
🟡CELL 3 — EDA: Class Distribution Plot
```

```python
os.makedirs('../screenshots', exist_ok=True)

fraud   = df[df['Class'] == 1]
genuine = df[df['Class'] == 0]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Class Distribution', fontsize=16, fontweight='bold')

counts = df['Class'].value_counts()
axes[0].bar(['Genuine','Fraud'], counts.values, color=['#2ECC71','#E74C3C'], edgecolor='black')
axes[0].set_title('Count')
axes[0].set_ylabel('Transactions')
for i, v in enumerate(counts.values):
    axes[0].text(i, v+1000, f'{v:,}', ha='center', fontweight='bold')

axes[1].pie(counts.values, labels=['Genuine','Fraud'],
            colors=['#2ECC71','#E74C3C'], autopct='%1.3f%%',
            explode=(0,0.15), shadow=True)
axes[1].set_title('Percentage')

plt.tight_layout()
plt.savefig('../screenshots/01_class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved!")
```

```python
🟡 CELL 4 — EDA: Correlation Heatmap
```

```python
fig, ax = plt.subplots(figsize=(16, 12))
corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, cmap='coolwarm', center=0,
            linewidths=0.3, ax=ax, annot=False)
ax.set_title('Feature Correlation Heatmap', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('../screenshots/02_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved!")
```

```python
🟡 CELL 5 — EDA: Amount & Time Analysis
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(genuine['Amount'], bins=100, alpha=0.6, color='#2ECC71',
             label='Genuine', density=True)
axes[0].hist(fraud['Amount'], bins=50, alpha=0.7, color='#E74C3C',
             label='Fraud', density=True)
axes[0].set_xlim(0, 2000)
axes[0].set_title('Amount Distribution (0–€2000)')
axes[0].set_xlabel('Amount (€)')
axes[0].legend()

df['Hour'] = (df['Time'] // 3600).astype(int)
fraud_hour   = df[df['Class']==1]['Hour'].value_counts().sort_index()
genuine_hour = df[df['Class']==0]['Hour'].value_counts().sort_index()
axes[1].bar(fraud_hour.index, fraud_hour.values, color='#E74C3C', alpha=0.8)
axes[1].set_title('Fraud Transactions by Hour')
axes[1].set_xlabel('Hour')

plt.tight_layout()
plt.savefig('../screenshots/03_amount_time.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved!")
```

```python
🟡 CELL 6 — Preprocessing
```

```python
# Remove duplicates
before = len(df)
df = df.drop_duplicates()
print(f"Duplicates removed: {before - len(df)}")

# Feature engineering
df['Log_Amount'] = np.log1p(df['Amount'])
df['Norm_Time']  = df['Time'] / df['Time'].max()

# Features aur target
X = df.drop(columns=['Class', 'Time', 'Amount'])
y = df['Class']

# Scale karo
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print(f"✅ Features: {X_scaled.shape}")
print(f"✅ Target:   {y.shape}")
print(f"Features list: {list(X.columns)}")
```

```python
🟡 CELL 7 — SMOTE
```

```python
print(f"Before SMOTE: {Counter(y)}")

smote = SMOTE(random_state=42, k_neighbors=5)
X_final, y_final = smote.fit_resample(X_scaled, y)

print(f"After  SMOTE: {Counter(y_final)}")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(['Genuine','Fraud'], [Counter(y)[0], Counter(y)[1]],
            color=['#2ECC71','#E74C3C'])
axes[0].set_title('Before SMOTE')

axes[1].bar(['Genuine','Fraud'], [Counter(y_final)[0], Counter(y_final)[1]],
            color=['#2ECC71','#E74C3C'])
axes[1].set_title('After SMOTE')

plt.tight_layout()
plt.savefig('../screenshots/04_smote_balance.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ SMOTE done!")
```

```python
🟡 CELL 8 — Train Test Split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y_final,
    test_size=0.2,
    random_state=42,
    stratify=y_final
)

print(f"✅ X_train: {X_train.shape}")
print(f"✅ X_test:  {X_test.shape}")
print(f"Train fraud %: {y_train.mean()*100:.2f}%")
print(f"Test  fraud %: {y_test.mean()*100:.2f}%")
```

```python
🟡 CELL 9 — Train All Models
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import time

models = {
    'Logistic Regression': LogisticRegression(
        max_iter=200,
        random_state=42,
        n_jobs=-1
    ),
    'Decision Tree': DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    ),
    'Random Forest': RandomForestClassifier(
        n_estimators=30,     # ← 100 ki jagah 30
        max_depth=8,
        random_state=42,
        n_jobs=-1
    ),
    'XGBoost': XGBClassifier(
        n_estimators=30,     # ← 100 ki jagah 30
        max_depth=4,
        learning_rate=0.1,
        random_state=42,
        verbosity=0,
        eval_metric='logloss',
        n_jobs=-1
    ),
    'LightGBM': LGBMClassifier(
        n_estimators=30,     # ← 100 ki jagah 30
        max_depth=4,
        learning_rate=0.1,
        random_state=42,
        verbosity=-1,
        n_jobs=-1
    ),
}

trained_models = {}

for name, model in models.items():
    print(f"⏳ {name}...", end=' ', flush=True)
    t0 = time.time()
    model.fit(X_train, y_train)
    trained_models[name] = model
    print(f"✅ {time.time()-t0:.1f}s")

print("\n🎉 Saare models train ho gaye!")
```

```python
🟡 CELL 10 — Evaluate All Models
```

```python
results = {}
for name, model in trained_models.items():
    yp    = model.predict(X_test)
    yprob = model.predict_proba(X_test)[:,1]
    results[name] = {
        'Accuracy':  round(accuracy_score(y_test, yp), 4),
        'Precision': round(precision_score(y_test, yp), 4),
        'Recall':    round(recall_score(y_test, yp), 4),
        'F1 Score':  round(f1_score(y_test, yp), 4),
        'ROC AUC':   round(roc_auc_score(y_test, yprob), 4),
    }

results_df = pd.DataFrame(results).T.sort_values('F1 Score', ascending=False)
print(results_df)

best_name = results_df['F1 Score'].idxmax()
print(f"\n🏆 Best Model: {best_name}")
```

```python
🟡 CELL 11 — Confusion Matrix Plot
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Confusion Matrices', fontsize=16, fontweight='bold')

for ax, (name, model) in zip(axes.flatten(), trained_models.items()):
    yp = model.predict(X_test)
    cm = confusion_matrix(y_test, yp)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Genuine','Fraud'],
                yticklabels=['Genuine','Fraud'])
    f1  = f1_score(y_test, yp)
    auc = roc_auc_score(y_test, model.predict_proba(X_test)[:,1])
    ax.set_title(f'{name}\nF1={f1:.3f}  AUC={auc:.3f}')
    ax.set_xlabel('Predicted')
    ax.set_ylabel('Actual')

plt.tight_layout()
plt.savefig('../screenshots/05_confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved!")
```

```python
🟡 CELL 12 — ROC Curve
```

```python
fig, ax = plt.subplots(figsize=(10, 7))
colors = ['#3498DB','#E74C3C','#2ECC71','#F39C12','#9B59B6','#1ABC9C']

for (name, model), color in zip(trained_models.items(), colors):
    yprob = model.predict_proba(X_test)[:,1]
    fpr, tpr, _ = roc_curve(y_test, yprob)
    auc = roc_auc_score(y_test, yprob)
    ax.plot(fpr, tpr, color=color, linewidth=2, label=f'{name} (AUC={auc:.3f})')

ax.plot([0,1],[0,1],'k--', label='Random')
ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curves - All Models', fontsize=14, fontweight='bold')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../screenshots/06_roc_curves.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Plot saved!")
```

```python
🟡 CELL 13 — Hyperparameter Tuning
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

rf_params = {
    'n_estimators': [30, 50],        # ← kam options
    'max_depth':    [5, 8, 10],
    'min_samples_split': [2, 5],
}

rf_search = RandomizedSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    rf_params,
    n_iter=5,          # ← 15 ki jagah sirf 5
    scoring='f1',
    cv=2,              # ← 3 ki jagah 2
    random_state=42,
    n_jobs=-1,
    verbose=1
)
rf_search.fit(X_train, y_train)
best_model = rf_search.best_estimator_

print("Best Params:", rf_search.best_params_)
yp_best = best_model.predict(X_test)
print(f"F1  After Tuning: {f1_score(y_test, yp_best):.4f}")
print(f"AUC After Tuning: {roc_auc_score(y_test, best_model.predict_proba(X_test)[:,1]):.4f}")
```

```python
🟡 CELL 14 — Feature Importance
```

```python
importances = best_model.feature_importances_
feat_df = pd.DataFrame({'Feature': X.columns, 'Importance': importances})
feat_df = feat_df.sort_values('Importance', ascending=False).head(15)

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(feat_df['Feature'][::-1], feat_df['Importance'][::-1], color='#E74C3C')
ax.set_title('Top 15 Feature Importances', fontsize=14, fontweight='bold')
ax.set_xlabel('Importance Score')
plt.tight_layout()
plt.savefig('../screenshots/07_feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()

print("Top 5 Features:")
print(feat_df.head())
```

```python
🟡 CELL 15 — Save Model ⭐ (Sabse Zaroori)
```

```python
os.makedirs('../models', exist_ok=True)

joblib.dump(best_model,           '../models/fraud_model.pkl')
joblib.dump(scaler,               '../models/scaler.pkl')
joblib.dump(list(X.columns),      '../models/feature_names.pkl')

print("✅ Models saved successfully!")
print("   models/fraud_model.pkl")
print("   models/scaler.pkl")
print("   models/feature_names.pkl")

# Verify
m = joblib.load('../models/fraud_model.pkl')
print(f"\n✅ Verify: Model loaded back — {type(m).__name__}")
```
